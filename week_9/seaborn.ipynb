```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("/workspaces/data_5500_code/week_9/tips.csv")

sns.lineplot(x="sex", y="total_bill", data=data)

# set title using matplotlib
plt.title("Line Plot of how Much People Spend")

plt.show()
```

Scatterplot

```python
sns.scatterplot(x='day', y='tip', data=data,)
plt.show()
```

```python
sns.scatterplot(x='day', y='tip', data=data, hue="sex")
plt.show()
```

Line Chart

```python
sns.lineplot(x="day", y="tip", data=data)
plt.show()
```

Chart Using Only One Attribute

```python
sns.lineplot(data=data.drop(['total_bill'], axis=1))
plt.show()
```

Bar Chart

```python
sns.barplot(x='day',y='tip', data=data, hue='sex')
plt.show()
```

Histogram

```python
sns.histplot(x='total_bill', data=data, kde=True, hue='sex')
plt.show()
```
