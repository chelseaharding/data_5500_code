Boolean Indexing

```python
import pandas as pd

df = pd.read_csv("/workspaces/data_5500_code/week_7/titanic.csv")
df.head()
```

```python
survived = df["Survived"] == 1 # mask
# survived

survivors = df[df["Survived"] == 1]
survivors
```

```python
# do not use the standard python boolean key words
rich_survivors = df[(df["Survived"] == 1) & (df["Pclass"] == 1)]
rich_survivors

# pandas uses & for and, | for or, ~ for not
```

```python
expensive_or_first = df[(df["Fare"] > 200) | (df["Pclass"] == 1)]
expensive_or_first
```

```python
# isin()

ports = df[df["Embarked"].isin(["C", "Q"])]
ports
```
