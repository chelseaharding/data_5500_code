Handling messy Data

```python
import pandas as pd

df = df = pd.read_csv("/workspaces/data_5500_code/week_7/titanic.csv")
df.head()
```

```python
df.isna().sum()
```

```python
# 2. Dropping Missing Data (.dropna)
# By default, dropna() drops ANY row that has at least one missing value anywhere.
# Let's check how many rows that leaves us with (spoiler: it's brutal).
print("Original shape:", df.shape)
clean_dropped = df.dropna()
print("Shape after dropping all rows with any NaN:", clean_dropped.shape)
```

```python
# A smarter way to drop: target specific columns
# Only drop rows where 'Embarked' is missing (which is only 2 rows in Titanic)
embarked_clean = df.dropna(subset=['Embarked'])
print(embarked_clean.shape)
```

```python
# 3. Filling Missing Data (.fillna)
# Let's look at the Age column's missing values
print("Missing ages before:", df['Age'].isna().sum())

# What should we fill them with? The median age is usually safer than the mean 
# because age can be skewed by outliers.
median_age = df['Age'].median()
print("Median Age:", median_age)

# Fill the missing ages with the median
df['Age_filled'] = df['Age'].fillna(median_age)

# Verify they are gone!
print("Missing ages after:", df['Age_filled'].isna().sum())
```

```python
# Filling categorical text data
# If a cabin is missing, let's label it "Unknown"
df['Cabin'] = df['Cabin'].fillna('Unknown')
print(df['Cabin'].head(5))
```
