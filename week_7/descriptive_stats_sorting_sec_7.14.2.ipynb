Descriptive Statistics

```python
import pandas as pd

df = pd.read_csv("/workspaces/data_5500_code/week_7/titanic.csv")
df.head()
```

```python
df.describe()
```

```python
print("Average age:", df["Age"].mean())

print("Total survivors:", df["Survived"].sum())

print("Unique ports:", df["Embarked"].unique())
```

Sorting DataFrames

```python
expensive_passengers = df.sort_values(by="Fare", ascending=False)
expensive_passengers[["Name", "Fare", "Pclass"]]
```

```python
sorted_titanic = df.sort_values(by=["Pclass", "Name"], ascending=[True, True])
sorted_titanic[["Name", "Pclass", "Age"]]
```
